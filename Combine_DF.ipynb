```python
import pandas as pd
df1=pd.read_csv(r"D:/BCA & MSC Core Notes/MSC/MSC 2ND YEAR/MSC Major project/Dataset/Berkeley.csv")
display(df1.head(5))
df2=pd.read_csv(r"D:/BCA & MSC Core Notes/MSC/MSC 2ND YEAR/MSC Major project/Dataset/CRU.csv")
display(df2.head(5))
df3=pd.read_csv(r"D:/BCA & MSC Core Notes/MSC/MSC 2ND YEAR/MSC Major project/Dataset/NASA.csv")
display(df3.head(5))
df4=pd.read_csv(r"D:/BCA & MSC Core Notes/MSC/MSC 2ND YEAR/MSC Major project/Dataset/NOAA.csv")
display(df4.head(5))
```

```python
import numpy as np

# Standardize column names and convert 'Date' to datetime objects for each DataFrame

df1.columns = ['Date', 'Berkeley_Temperature']
df1['Date'] = pd.to_datetime(df1['Date'])
df1['Berkeley_Temperature'] = pd.to_numeric(df1['Berkeley_Temperature'], errors='coerce')

df2.columns = ['Date', 'CRU_Temperature']
df2['Date'] = pd.to_datetime(df2['Date'])
df2['CRU_Temperature'] = pd.to_numeric(df2['CRU_Temperature'], errors='coerce')

df3.columns = ['Date', 'NASA_Temperature']
df3['Date'] = pd.to_datetime(df3['Date'])
df3['NASA_Temperature'] = pd.to_numeric(df3['NASA_Temperature'], errors='coerce')

df4.columns = ['Date', 'NOAA_Temperature']
df4['Date'] = pd.to_datetime(df4['Date'])
df4['NOAA_Temperature'] = pd.to_numeric(df4['NOAA_Temperature'], errors='coerce')

# Merge the dataframes on the 'Date' column
combined_df = pd.merge(df1, df2, on='Date', how='outer')
combined_df = pd.merge(combined_df, df3, on='Date', how='outer')
combined_df = pd.merge(combined_df, df4, on='Date', how='outer')

combined_df.to_csv(r"D:\BCA & MSC Core Notes\MSC\MSC 2ND YEAR\MSC Major project\Dataset\combined_df.csv")
```
